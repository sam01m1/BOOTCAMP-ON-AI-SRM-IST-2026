```python
import random

numbers = [random.randint(1, 100) for _ in range(10)]

print("Random numbers:")
print(numbers)

print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
text = "Python is easy to learn"

print("Original string:")
print(text)

print("Reversed string:")
print(text[::-1])

words = text.split()
print("Words in reverse order:")
print(" ".join(words[::-1]))
```
